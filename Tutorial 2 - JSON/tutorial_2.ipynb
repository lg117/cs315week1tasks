## 1. Writing JSON with Python

```python
import json
food_ratings = {"organic dog food": 2, "human food": 10}
json.dumps(food_ratings)


numbers_present = {1: True, 2: True, 3: False}
json.dumps(numbers_present)
```

```python
import json

cat_data = {
  "name": "napo",
  "is_dog": False,
  "hobbies": ["eating", "sleeping", "meowing",],
  "age": 10,
  "address": {
    "work": "biscuit making",
    "home": ("Alexandria", "Virginia",),
  },
  "friends": [
    {
      "name": "Freddie",
      "hobbies": ["eating", "sleeping", "sprinting",],
    },
  ],
}

with open("hello_napo.json", mode="w", encoding="utf-8") as write_file:
    json.dump(cat_data, write_file)
```

```python
import json
cat_registry = {1: {"name": "napo"}}
cat_json = json.dumps(cat_registry)
cat_json
```

```python
import json
with open("hello_napo.json", mode="r", encoding="utf-8") as read_file:
    naps_data = json.load(read_file)

type(naps_data)


naps_data["name"]
```

```python
from hello_napo import cat_data
naps_data == cat_data


type(naps_data["address"]["home"])


type(cat_data["address"]["home"])
```
